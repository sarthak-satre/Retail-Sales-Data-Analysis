```python
# 1...calculate total revenue , this is the gross sales revenue not ater discounts..
df['Sales']=df['Quantity'] * df['Unit_Price']
```

```python
df['Sales'].sum()
```

```python
# 2...calculate how much discount the company gives on its sales.
df['Discount'] = df['Sales'] * df['Discount_Percent'] / 100
```

```python
df['Discount'].sum()
```

```python
# 3.net sales after minus
df['Net_Sales']=df['Sales']-df['Discount']
```

```python
df['Net_Sales'].sum()
```

```python
# product generates the highest sales.
df.groupby('Product')['Net_Sales'].sum().sort_values(ascending=False)
```

```python
# category wise sales
df.groupby('Category')['Net_Sales'].sum().sort_values(ascending=False)
```

```python
# city wise sales
df.groupby('City')['Net_Sales'].sum().sort_values(ascending=False)
```

```python
# region wise sales
df.groupby('Region')['Net_Sales'].sum().sort_values(ascending=False)
```

```python
# month wise
df['Month']=df['Order_Date'].dt.month
df
```

```python
# sales month wise
df.groupby('Month')['Net_Sales'].sum()
```

```python
import matplotlib.pyplot as plt

df.groupby('Month')['Net_Sales'].sum().plot(kind='bar')

plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Net Sales')
plt.show()
```
